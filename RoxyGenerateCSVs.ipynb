```python
import csv
import random
from datetime import datetime, timedelta

# Set random seed for consistent data generation
random.seed(42)

# Configuration parameters
NUM_ACCOUNTS = 500
NUM_CONTRACTS = 1000

# -------------------------------------------------------------------
# 1. GENERATE DIMENSION: account_demographics.csv
# -------------------------------------------------------------------
regions = ['MIDWEST', 'Midwest ', 'mw', 'West', 'WEST', 'East', 'EAST', '']
risk_tiers = ['Low', 'LOW', 'Med', 'High', 'high', '']
acc_open_dates = ['2025-11-28', '2025-10-01', '2025-03-25', '00/00/2025', '2025-05-19']

with open('account_demographics.csv', mode='w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['account_id', 'region', 'risk_tier', 'account_open_date'])
    for i in range(1000, 1000 + NUM_ACCOUNTS):
        writer.writerow([
            f"ACC-{i}",
            random.choice(regions),
            random.choice(risk_tiers),
            random.choice(acc_open_dates)
        ])

# -------------------------------------------------------------------
# 2. GENERATE FACT TABLE: contract_ledger.csv
# -------------------------------------------------------------------
payment_statuses = ['Past Due', 'Current', 'past due', ' Paid ', 'CURRENT', 'Paid']
vendors = ['VNDR_A', 'vndr_a', 'VNDR_B', 'vndr_b', 'VNDR_C', 'vndr_c']
products = ['Fleet_Car', 'FLEET_TRUCK', 'Equipment']
pay_methods = ['ACH', 'Credit Card', 'Autopay ', 'Autopay']
delinquency_cycles = ['0', '30', '60', '90', '']
underwriters = ['UW_99', 'UW_102', 'UW_45']

with open('contract_ledger.csv', mode='w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow([
        'contract_id', 'account_id', 'total_contract_value', 
        'monthly_payment_amount', 'payment_status', 'vendor_code', 
        'product_category', 'payment_method', 'delinquency_cycle', 
        'underwriter_id', 'collateral_value', 'interest_rate'
    ])
    
    for i in range(5000, 5000 + NUM_CONTRACTS):
        # Intentionally inject occasional dollar signs to dirty the values
        raw_val = random.randint(15000, 150000)
        total_val = f"${raw_val}" if random.random() < 0.2 else str(raw_val)
        
        # Intentionally inject 'k' formatting into collateral values
        raw_coll = random.randint(20, 160)
        coll_val = f"{raw_coll}k" if random.random() < 0.5 else str(raw_coll * 1000)
        
        # Intentionally cross-contaminate interest rate formats
        raw_rate = round(random.uniform(0.04, 0.085), 3)
        rate_val = f"{round(raw_rate * 100, 1)}%" if random.random() < 0.5 else str(raw_rate)
        
        # Pick random accounts, intentionally allowing duplicates for duplicates trap
        acc_id = f"ACC-{random.randint(1000, 1000 + NUM_ACCOUNTS - 1)}"
        
        writer.writerow([
            f"CON-{i}", acc_id, total_val, random.randint(800, 4500),
            random.choice(payment_statuses), random.choice(vendors),
            random.choice(products), random.choice(pay_methods),
            random.choice(delinquency_cycles), random.choice(underwriters),
            coll_val, rate_val
        ])

# -------------------------------------------------------------------
# 3. GENERATE DIMENSION: operational_logs.csv
# -------------------------------------------------------------------
stages = ['Stage 1', 'Stage 2', 'Stage 3', 'Stage 4', '1', 'Unknown']

with open('operational_logs.csv', mode='w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['log_id', 'contract_id', 'current_stage', 'days_delayed_in_stage', 'last_ping_timestamp'])
    
    start_date = datetime(2026, 1, 1)
    for i in range(8000, 8000 + NUM_CONTRACTS):
        # Inject -999 dummy value outliers
        delay = -999 if random.random() < 0.1 else random.randint(0, 45)
        ping_time = start_date + timedelta(days=random.randint(0, 200), hours=random.randint(0, 23))
        
        writer.writerow([
            f"LOG-{i}",
            f"CON-{i - 3000}", # Maps back cleanly to corresponding contract IDs
            random.choice(stages),
            delay,
            ping_time.strftime('%Y-%m-%d %H:%M:%S')
        ])

print("Successfully generated 3 messy pipeline tables: account_demographics.csv, contract_ledger.csv, and operational_logs.csv!")

```
